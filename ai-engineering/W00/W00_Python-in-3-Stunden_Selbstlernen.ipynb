```python
# Hilfsfunktion für die Selbsttests – einfach ausführen, nicht ändern.
import math

def check(name, test):
    try:
        ok = bool(test())
    except Exception as e:
        print(f"❌ {name}: noch nicht gelöst ({type(e).__name__}: {e})")
        return
    print(f"✅ {name}" if ok else f"❌ {name}: Ergebnis stimmt noch nicht")

print("Selbsttests bereit.")
```

## 1 · Python als Taschenrechner

| Operator | Bedeutung | Beispiel | Ergebnis |

## 4 · Bedingungen

`if` führt Code nur aus, wenn eine Bedingung wahr ist. `elif` prüft weitere Fälle, `else` fängt den Rest ab.
Der Code im Block ist **eingerückt** (4 Leerzeichen). Die Einrückung ist in Python Pflicht, sie zeigt, was zusammengehört.

Vergleiche: `==` (gleich), `!=` (ungleich), `<`, `<=`, `>`, `>=`. Verknüpfung mit `and`, `or`, `not`.

```python
temperatur = 66
if temperatur > 80:
    status = "rot"
elif temperatur > 60:
    status = "gelb"
else:
    status = "grün"
print(f"{temperatur} °C → {status}")
```

## 5 · Schleifen

Eine `for`-Schleife wiederholt einen Block für jedes Element einer Liste.
`range(n)` erzeugt die Zahlen 0 bis n−1, `range(a, b)` die Zahlen von a bis b−1.

Ein häufiges Muster ist der **Akkumulator**: einen Startwert festlegen und in jedem Durchlauf aktualisieren.

```python
temperaturen = [58, 61, 64, 70, 66, 72]
summe = 0
for t in temperaturen:
    summe = summe + t
print("Summe:", summe, "| Mittelwert:", summe / len(temperaturen))

for i in range(3):
    print("Durchlauf", i)
```

## 7 · Fehlermeldungen lesen

Fehler gehören zum Programmieren dazu. Python zeigt dann einen **Traceback**. Lest ihn **von unten**: Die letzte Zeile nennt Fehlertyp und Grund.

| Fehler | Typische Ursache |
|---|---|
| `NameError` | Variable falsch geschrieben oder Zelle mit der Definition nicht ausgeführt |
| `SyntaxError` | Tippfehler im Aufbau, zum Beispiel fehlender Doppelpunkt oder fehlende Klammer |
| `IndentationError` | Einrückung stimmt nicht |
| `TypeError` | Unpassende Typen, zum Beispiel Text + Zahl |
| `ZeroDivisionError` | Division durch null |

⚠️ Die nächsten zwei Zellen erzeugen **absichtlich** Fehler. Führt sie aus, lest die letzte Zeile und repariert sie dann selbst.

```python
stueckzahl = 45
print(stuekzahl)
```

```python
ausgabe = "Teile: " + 45
print(ausgabe)
```

<details><summary>Lösung anzeigen</summary>

1. `NameError`: Der Name ist falsch geschrieben, richtig ist `stueckzahl`.
2. `TypeError`: Text und Zahl lassen sich nicht mit `+` verbinden. Richtig: `"Teile: " + str(45)` oder `f"Teile: {45}"`.
</details>

```python
from google.colab import userdata
from openai import OpenAI

client = OpenAI(api_key=userdata.get("KIT_TOOLBOX_KEY").strip(), base_url="https://ki-toolbox.scc.kit.edu/api/v1")

antwort = client.chat.completions.create(
    model="kit.mistral-small-4-119b-a8b",
    max_tokens=100,
    messages=[{"role": "user", "content": "Begrüße in einem Satz die Teilnehmenden eines AI-Kurses am KIT."}],
)

print(antwort.choices[0].message.content)
print(f"Verbrauch: {antwort.usage.prompt_tokens} Input-Tokens, {antwort.usage.completion_tokens} Output-Tokens")
```
